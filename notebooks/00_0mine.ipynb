```python
from graphviz import Digraph

import os

# Ścieżka do katalogu z dot.exe (dostosuj jeśli potrzebujesz)
graphviz_path = r"C:\Program Files\Graphviz\bin"
dot_exe = os.path.join(graphviz_path, "dot.exe")

# Ustaw zmienną dla Graphviz
os.environ["GRAPHVIZ_DOT"] = dot_exe

# Opcjonalnie: dodaj do PATH (na wszelki wypadek)
if graphviz_path not in os.environ.get("PATH", ""):
    os.environ["PATH"] += os.path.pathsep + graphviz_path


dot = Digraph(
    "proces_uczenia_modelu",
    format="png"
)

# Ustawienia ogólne
dot.attr(
    rankdir="TB",
    bgcolor="white",
    fontname="Arial",
    pad="0.3",
    nodesep="0.45",
    ranksep="0.6"
)

dot.attr(
    "node",
    shape="box",
    style="rounded",
    fontname="Arial",
    fontsize="12",
    margin="0.25,0.20"
)

dot.attr(
    "edge",
    fontname="Arial",
    fontsize="10"
)


# ==========================================================
# 1. DANE WEJŚCIOWE
# ==========================================================

dot.node(
    "data",
    "Zbiór danych obrazowych"
)

dot.node(
    "split",
    "Podział danych"
)

dot.node(
    "train",
    "Zbiór treningowy"
)

dot.node(
    "val",
    "Zbiór walidacyjny"
)

dot.node(
    "test",
    "Zbiór testowy"
)

dot.edge("data", "split")

dot.edge("split", "train")
dot.edge("split", "val")
dot.edge("split", "test")


# ==========================================================
# 2. PRZYGOTOWANIE DANYCH TRENINGOWYCH
# ==========================================================

dot.node(
    "preprocessing_train",
    "Wstępne przetwarzanie\n"
    "• zmiana rozmiaru obrazu\n"
    "• normalizacja\n"
    "• przygotowanie danych"
)

dot.node(
    "augmentation",
    "Augmentacja danych\n"
    "• odbicie\n"
    "• zmiana jasności i kontrastu\n"
    "• rozmycie\n"
    "• częściowe zasłonięcie"
)

dot.edge(
    "train",
    "preprocessing_train"
)

dot.edge(
    "preprocessing_train",
    "augmentation"
)


# ==========================================================
# 3. PRZYGOTOWANIE WALIDACJI I TESTU
# ==========================================================

dot.node(
    "preprocessing_val",
    "Wstępne przetwarzanie\n"
    "zbioru walidacyjnego"
)

dot.node(
    "preprocessing_test",
    "Wstępne przetwarzanie\n"
    "zbioru testowego"
)

dot.edge(
    "val",
    "preprocessing_val"
)

dot.edge(
    "test",
    "preprocessing_test"
)


# ==========================================================
# 4. MODEL
# ==========================================================

dot.node(
    "pretrained",
    "Model wstępnie wytrenowany\n"
    "(uczenie transferowe)"
)

dot.node(
    "model",
    "Dostrajanie modelu"
)

dot.edge(
    "pretrained",
    "model"
)

dot.edge(
    "augmentation",
    "model"
)


# ==========================================================
# 5. PROCES UCZENIA
# ==========================================================

dot.node(
    "batch",
    "Partia danych\n"
)

dot.node(
    "prediction",
    "Predykcja modelu"
)

dot.node(
    "loss",
    "Obliczenie funkcji straty\n"
)

dot.node(
    "gradient",
    "Wyznaczenie gradientu"
)

dot.node(
    "optimizer",
    "Aktualizacja parametrów\n"
    "przez optymalizator"
)

dot.edge(
    "model",
    "batch"
)

dot.edge(
    "batch",
    "prediction"
)

dot.edge(
    "prediction",
    "loss"
)

dot.edge(
    "loss",
    "gradient"
)

dot.edge(
    "gradient",
    "optimizer"
)

dot.edge(
    "optimizer",
    "batch",
    label="kolejna iteracja"
)


# ==========================================================
# 6. EPOKA + WALIDACJA
# ==========================================================

dot.node(
    "epoch",
    "Zakończenie epoki"
)

dot.node(
    "validation",
    "Ocena na zbiorze walidacyjnym\n"
    "• strata walidacyjna\n"
    "• jakość predykcji"
)

dot.node(
    "repeat",
    "Kolejna epoka?"
)

dot.edge(
    "optimizer",
    "epoch"
)

dot.edge(
    "epoch",
    "validation"
)

dot.edge(
    "preprocessing_val",
    "validation"
)

dot.edge(
    "validation",
    "repeat"
)

dot.edge(
    "repeat",
    "batch",
    label="TAK"
)

dot.edge(
    "repeat",
    "best_model",
    label="NIE"
)


# ==========================================================
# 7. WYBÓR NAJLEPSZEGO MODELU
# ==========================================================

dot.node(
    "best_model",
    "Wybór najlepszego modelu\n"
    "na podstawie walidacji"
)

dot.node(
    "test_model",
    "Ocena końcowa\n"
    "na zbiorze testowym"
)

dot.edge(
    "best_model",
    "test_model"
)

dot.edge(
    "preprocessing_test",
    "test_model"
)


# ==========================================================
# 8. KOŃCOWA OCENA
# ==========================================================

dot.node(
    "metrics",
    "Metryki jakości modelu\n"
    "• Accuracy\n"
    "• Precision\n"
    "• Recall\n"
    "• F1-score\n"
    "• czas inferencji"
)

dot.node(
    "analysis",
    "Analiza wyników i błędów"
)

dot.edge(
    "test_model",
    "metrics"
)

dot.edge(
    "metrics",
    "analysis"
)


# ==========================================================
# GENEROWANIE PLIKU
# ==========================================================

output_path = dot.render(
    filename="proces_uczenia_modelu",
    directory=r"C:\Users\klanz\Desktop\MAGISTERKA\notebooks\wyniki1",
    cleanup=True
)

print(f"Schemat zapisano jako:\n{output_path}")
```
