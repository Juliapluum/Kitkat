```python
DATASET = Path(r"C:\\Users\\klanz\\Desktop\\MAGISTERKA\\dataset")

print(DATASET.exists())
```

```python
yaml_text = f"""
path: {DATASET}

train: train/images
val: val/images
test: test/images

names:
    0: chicken
    1: not_chicken
"""

with open(DATASET/"dataset.yaml","w") as f:
    f.write(yaml_text)

print("dataset.yaml utworzony.")
```

```python
print((DATASET/"train/images").exists())
print((DATASET/"train/labels").exists())

print((DATASET/"val/images").exists())
print((DATASET/"val/labels").exists())

print((DATASET/"test/images").exists())
print((DATASET/"test/labels").exists())
```

```python
model = YOLO("yolov8s.pt")
```

```python
results = model.train(

    data=DATASET/"dataset.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    workers=8,
    patience=20,
    project="runs",
    name="YOLOv8_baseline",
    pretrained=True,
    verbose=True,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=0,
    translate=0.05,
    scale=0.15,
    shear=0,
    perspective=0,
    flipud=0,
    fliplr=0.5,
    mosaic=0.5,
    mixup=0,
    copy_paste=0,
    erasing=0,

)
```

```python
import albumentations as A
from ultralytics import YOLO

import torch
from pathlib import Path

model7 = YOLO("yolov8s.pt")
DATASET = Path("C:\\Users\\klanz\\Desktop\\MAGISTERKA\\dataset")
custom_transforms = [
    A.MotionBlur(blur_limit=(3, 15), p=0.3),  # p = probability of application
    A.RandomBrightnessContrast(brightness_limit=(-0.5,0), contrast_limit=(-0.5,0), p=0.5)
]
```

```python
results7 = model7.train(

    data=DATASET/"dataset.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    workers=8,
    patience=20,
    project="runs",
    name="YOLOv8_baseline",
    pretrained=True,
    verbose=True,
    hsv_h=0,
    hsv_s=0,
    hsv_v=0,
    degrees=0,
    translate=0.05,
    scale=0.15,
    shear=0,
    perspective=0,
    flipud=0,
    fliplr=0.5,
    mosaic=0.5,
    mixup=0,
    copy_paste=0,
    erasing=0.3,
    augmentations=custom_transforms


)
```
